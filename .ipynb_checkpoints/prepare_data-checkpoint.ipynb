```python
import numpy as np
from sklearn.model_selection import train_test_split
from datetime import datetime
import pandas as pd

data = pd.read_csv('data/iris.csv')

data["iris_id"] = range(len(data))
data["event_timestamp"] = datetime.now()

data.to_parquet("feature_repo/data/iris.parquet", index=False)

train, test = train_test_split(data, test_size = 0.4, stratify = data['species'], random_state = 42)

train.to_parquet("feature_repo/data/iris_train.parquet", index=False)
test.to_parquet("feature_repo/data/iris_eval.parquet", index=False)

print("Train:", train.shape)
print("Eval:", test.shape)
```
